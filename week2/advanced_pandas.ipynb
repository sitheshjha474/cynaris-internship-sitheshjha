# Merge DataFrames — All Join Types

```python
import pandas as pd

sales = pd.read_csv("da_sales_transactions.csv")
customers = pd.read_csv("da_customer_data.csv")

# Inner Join
inner = pd.merge(sales, customers, on="Customer_ID", how="inner")

# Left Join
left = pd.merge(sales, customers, on="Customer_ID", how="left")

# Right Join
right = pd.merge(sales, customers, on="Customer_ID", how="right")

# Outer Join
outer = pd.merge(sales, customers, on="Customer_ID", how="outer")

inner.head()
```

# Pivot Table — Sales by Region and Product

```python
import pandas as pd

df = pd.read_csv("da_sales_transactions.csv")

df["Sales"] = df["Units"] * df["Unit_Price"]

pivot = pd.pivot_table(
    df,
    values="Sales",
    index="City",
    columns="Category",
    aggfunc="sum"
)

pivot
```

# pd.melt() — Wide to Long Format

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["A", "B", "C"],
    "Jan": [100, 150, 200],
    "Feb": [120, 170, 220],
    "Mar": [140, 180, 250]
})

long_df = pd.melt(
    df,
    id_vars="Product",
    var_name="Month",
    value_name="Sales"
)

long_df
```

# Method Chaining

```python
sales = pd.read_csv("da_sales_transactions.csv")

result = (
    sales
    .assign(Sales=lambda x: x["Units"] * x["Unit_Price"])
    .groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

result
```
