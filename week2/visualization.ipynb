```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("da_sales_transactions.csv")

sns.set_theme(style="whitegrid")

df["Sales"] = df["Units"] * df["Unit_Price"]

df.groupby("Date")["Sales"].sum().plot(figsize=(10,5))

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.tight_layout()
plt.savefig("line_chart.png")
plt.show()
```

Bar Chart — Sales by Category

We'll use the actual Category, Units, and Unit_Price columns.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("da_sales_transactions.csv")

sns.set_theme(style="whitegrid")

df["Sales"] = df["Units"] * df["Unit_Price"]

df.groupby("Category")["Sales"].sum().plot(kind="bar", figsize=(8,5))

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.tight_layout()
plt.savefig("bar_chart.png")
plt.show()
```

# Scatter Chart — Units vs Unit Price

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("da_sales_transactions.csv")

sns.set_theme(style="whitegrid")

plt.scatter(df["Units"], df["Unit_Price"])

plt.title("Units vs Unit Price")
plt.xlabel("Units")
plt.ylabel("Unit Price")

plt.tight_layout()
plt.savefig("scatter_chart.png")
plt.show()
```

# Histogram — Unit Price Distribution

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("da_sales_transactions.csv")

sns.set_theme(style="whitegrid")

plt.hist(df["Unit_Price"], bins=10)

plt.title("Distribution of Unit Price")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")

plt.tight_layout()
plt.savefig("histogram.png")
plt.show()
```

# Box Chart — Unit Price

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("da_sales_transactions.csv")

sns.set_theme(style="whitegrid")

plt.boxplot(df["Unit_Price"])

plt.title("Unit Price Distribution")
plt.ylabel("Unit Price")

plt.tight_layout()
plt.savefig("box_chart.png")
plt.show()
```
