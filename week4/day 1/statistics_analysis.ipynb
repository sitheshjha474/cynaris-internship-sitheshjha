```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("da_sales_transactions.csv")

df.head()
```

# Mean, Median, Mode, Variance, Standard Deviation

```python
x = df["Unit_Price"]

print("Mean:", x.mean())
print("Median:", x.median())
print("Mode:", x.mode()[0])
print("Variance:", x.var())
print("Standard Deviation:", x.std())
```

# Frequency Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Unit_Price"], bins=15, kde=True)

plt.title("Frequency Distribution of Unit Price")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

# 4. Percentiles and Quartiles

```python
print("25th Percentile (Q1):", x.quantile(0.25))
print("50th Percentile (Q2):", x.quantile(0.50))
print("75th Percentile (Q3):", x.quantile(0.75))
```

```python
# Quartile calculations
Q1 = x.quantile(0.25)
Q3 = x.quantile(0.75)

IQR = Q3 - Q1

print("IQR:", IQR)
```

# . Side-by-Side Box Plot

#Compare Online vs Offline prices:

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x="Channel", y="Unit_Price")

plt.title("Unit Price Distribution by Channel")
plt.xlabel("Channel")
plt.ylabel("Unit Price")
plt.tight_layout()
plt.show()
```
